# Examples of plotly-scala with Almond

These are the examples of the [plotly-scala demo](https://alexarchambault.github.io/plotly-scala/).

```scala
//> using dep org.plotly-scala::plotly-almond:0.8.6-SNAPSHOT

// restrict the output height to avoid scrolling in output cells
repl.pprinter() = repl.pprinter().copy(defaultHeight = 3)

// to have the plots available without an internet connection, call init(offline=true) after this import
import plotly._, plotly.element._, plotly.layout._, plotly.Almond._
```

# Line Charts
## Basic Line Plot

```scala
val trace1 = Scatter(
  Seq(1, 2, 3, 4),
  Seq(10, 15, 13, 17)
)

val trace2 = Scatter(
  Seq(1, 2, 3, 4),
  Seq(16, 5, 11, 9)
)

val data = Seq(trace1, trace2)

plot(data)
```

# Scatter Plots
## Categorical Dot Plot

```scala
val country = Seq(
  "Switzerland (2011)",
  "Chile (2013)",
  "Japan (2014)",
  "United States (2012)",
  "Slovenia (2014)",
  "Canada (2011)",
  "Poland (2010)",
  "Estonia (2015)",
  "Luxembourg (2013)",
  "Portugal (2011)"
)

val votingPop = Seq(
  40.0, 45.7, 52.0, 53.6, 54.1, 54.2, 54.5, 54.7, 55.1, 56.6
)

val regVoters = Seq(
  49.1, 42.0, 52.7, 84.3, 51.7, 61.1, 55.3, 64.2, 91.1, 58.9
)

val trace1 = Scatter(votingPop, country).updated(
  mode = ScatterMode(ScatterMode.Markers),
  name = "Percent of estimated voting age population",
  marker = Marker().updated(
    color = Color.RGBA(156, 165, 196, 0.95),
    line = Line().updated(
      color = Color.RGBA(156, 165, 196, 1.0),
      width = 1.0
    ),
    symbol = Symbol.Circle(),
    size = 16
  )
)

val trace2 = Scatter(regVoters, country).updated(
  mode = ScatterMode(ScatterMode.Markers),
  name = "Percent of estimated registered voters",
  marker = Marker().updated(
    color = Color.RGBA(204, 204, 204, 0.95),
    line = Line().updated(
      color = Color.RGBA(217, 217, 217, 1.0),
      width = 1.0
    ),
    symbol = Symbol.Circle(),
    size = 16
  )
)

val data = Seq(trace1, trace2)

val layout = Layout().updated(
  title = "Votes cast for ten lowest voting age population in OECD countries",
  xaxis = Axis().updated(
    showgrid = false,
    showline = true,
    linecolor = Color.RGB(102, 102, 102),
    titlefont = Font(Color.RGB(204, 204, 204)),
    tickfont = Font(Color.RGB(102, 102, 102)),
    autotick = false,
    dtick = 10.0,
    ticks = Ticks.Outside,
    tickcolor = Color.RGB(102, 102, 102)
  ),
  margin = Margin(
    l = 140,
    r = 40,
    b = 50,
    t = 80
  ),
  legend = Legend().updated(
    font = Font(size = 10),
    yanchor = Anchor.Middle,
    xanchor = Anchor.Right
  ),
  width = 600,
  height = 400,
  paper_bgcolor = Color.RGB(254, 247, 234),
  plot_bgcolor = Color.RGB(254, 247, 234),
  hovermode = HoverMode.Closest
)

plot(data, layout)
```

## Highlighted Points

```scala
// simulated weights and heights of 200 people
val random  = new scala.util.Random(8) // fixed seed, so that the plot is the same every time
val heights = Seq.fill(200)(170 + 9 * random.nextGaussian())
val weights = heights.map(h => 0.9 * h - 85 + 8 * random.nextGaussian())

// the people whose BMI is above 27 are selected, and highlighted
val bmis     = heights.zip(weights).map { case (h, w) => w / math.pow(h / 100, 2) }
val selected = bmis.zipWithIndex.collect { case (bmi, idx) if bmi > 27 => idx }

val data = Seq(
  Scatter(heights, weights).updated(
    mode = ScatterMode(ScatterMode.Markers),
    marker = Marker().updated(size = 8, color = Color.RGB(99, 110, 250)),
    selectedpoints = selected,
    selected = Selection().updated(
      marker = SelectionMarker().updated(color = Color.RGB(239, 85, 59), size = 11.0)
    ),
    unselected = Selection().updated(marker = SelectionMarker().updated(opacity = 0.3)),
    hovertemplate = "%{x:.0f} cm, %{y:.0f} kg<extra></extra>"
  )
)

val layout = Layout().updated(
  title = "People With a BMI Above 27",
  xaxis = Axis().updated(title = "height (cm)"),
  yaxis = Axis().updated(title = "weight (kg)")
)

plot(data, layout)
```

## Waterfall Bar Chart

```scala
val xData = Seq(
  "Product Revenue",
  "Services Revenue",
  "Total Revenue",
  "Fixed Costs",
  "Variable Costs",
  "Total Costs",
  "Total"
)

val yData = Seq(400, 660, 660, 590, 400, 400, 340)

val textList = Seq("$430K", "$260K", "$690K", "$-120K", "$-200K", "$-320K", "$370K")

// Base

val trace1 = Bar(xData, Seq(0, 430, 0, 570, 370, 370, 0)).updated(
  marker = Marker().updated(color = Color.RGBA(1, 1, 1, 0.0))
)

// Revenue

val trace2 = Bar(xData, Seq(430, 260, 690, 0, 0, 0, 0)).updated(
  marker = Marker().updated(
    color = Color.RGBA(55, 128, 191, 0.7),
    line = Line().updated(
      color = Color.RGBA(55, 128, 191, 1.0),
      width = 2.0
    )
  )
)

// Cost

val trace3 = Bar(xData, Seq(0, 0, 0, 120, 200, 320, 0)).updated(
  marker = Marker().updated(
    color = Color.RGBA(219, 64, 82, 0.7),
    line = Line().updated(
      color = Color.RGBA(219, 64, 82, 1.0),
      width = 2.0
    )
  )
)

// Profit

val trace4 = Bar(xData, Seq(0, 0, 0, 0, 0, 0, 370)).updated(
  marker = Marker().updated(
    color = Color.RGBA(50, 171, 96, 0.7),
    line = Line().updated(
      color = Color.RGBA(50, 171, 96, 1.0),
      width = 2.0
    )
  )
)

val data = Seq(trace1, trace2, trace3, trace4)

val annotations = xData.zip(yData).zip(textList).map { case ((x, y), text) =>
  Annotation().updated(
    x = x,
    y = y,
    text = text,
    font = Font(
      family = "Arial",
      size = 14,
      color = Color.RGBA(245, 246, 249, 1)
    ),
    showarrow = false
  )
}

val layout = Layout().updated(
  title = "Annual Profit 2015",
  barmode = BarMode.Stack,
  paper_bgcolor = Color.RGBA(245, 246, 249, 1),
  plot_bgcolor = Color.RGBA(245, 246, 249, 1),
  width = 600,
  height = 400,
  showlegend = false,
  annotations = annotations
)

plot(data, layout)
```

## Period Positioning

```scala
// monthly sales, and their running average: the values of each month are drawn in the middle
// of their month, rather than at its first day
val months  = (1 to 12).map(m => LocalDateTime(2025, m, 1, 0, 0, 0))
val sales   = Seq(42.0, 38.0, 51.0, 55.0, 61.0, 70.0, 74.0, 69.0, 58.0, 54.0, 63.0, 80.0)
val average = sales.scanLeft(0.0)(_ + _).tail.zipWithIndex.map { case (sum, i) => sum / (i + 1) }

val data = Seq(
  Bar(months, sales).updated(
    name = "sales",
    xperiod = "M1",
    xperiodalignment = PeriodAlignment.Middle
  ),
  Scatter(months, average).updated(
    name = "running average",
    mode = ScatterMode(ScatterMode.Lines, ScatterMode.Markers),
    xperiod = "M1",
    xperiodalignment = PeriodAlignment.Middle
  )
)

val layout = Layout().updated(
  title = "Monthly Sales",
  yaxis = Axis().updated(ticksuffix = "K"),
  xaxis = Axis().updated(tickformat = "%b")
)

plot(data, layout)
```

## Gradient And Pattern Fills

```scala
val hours = (0 to 24).map(_.toDouble)

// power produced by solar panels over a day, in kW, and the power consumed by a house
val produced = hours.map(h => math.max(0.0, 5 * math.sin((h - 6) * math.Pi / 13)))
val consumed =
  hours.map(h => 0.8 + 1.2 * math.exp(-math.pow(h - 8, 2) / 4) + 1.8 * math.exp(-math.pow(h - 19.5, 2) / 6))

val data = Seq(
  Scatter(hours, produced).updated(
    name = "produced",
    fill = Fill.ToZeroY,
    // the higher the power, the more saturated the fill
    fillgradient = FillGradient().updated(
      `type` = GradientType.Vertical,
      colorscale = ColorScale.CustomScale(
        Seq(0.0 -> Color.RGBA(254, 203, 82, 0.1), 1.0 -> Color.RGBA(254, 160, 40, 0.9))
      )
    ),
    line = Line().updated(color = Color.RGB(254, 160, 40), shape = LineShape.Spline)
  ),
  Scatter(hours, consumed).updated(
    name = "consumed",
    fill = Fill.ToZeroY,
    fillcolor = Color.RGBA(99, 110, 250, 0.15),
    fillpattern = FillPattern().updated(
      shape = PatternShape.DiagonalUp,
      fgcolor = Color.RGB(99, 110, 250),
      size = 8.0,
      solidity = 0.15
    ),
    line = Line().updated(color = Color.RGB(99, 110, 250), shape = LineShape.Spline)
  )
)

val layout = Layout().updated(
  title = "Solar Power Over a Day",
  xaxis = Axis().updated(title = "hour", dtick = 3.0),
  yaxis = Axis().updated(title = "kW")
)

plot(data, layout)
```

## Heatmap With Text

```scala
val days  = Seq("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
val hours = Seq("8h", "10h", "12h", "14h", "16h", "18h", "20h")

// average number of visitors of a museum, by day and hour
val visitors = hours.indices.map { h =>
  days.indices.map { d =>
    // busier towards the weekend
    val dayFactor = if (d >= 5) 1.8 else 0.7 + 0.12 * d
    val peak      = math.exp(-math.pow(h - 3.2, 2) / 4)
    math.round(dayFactor * (20 + 110 * peak) * (if (d == 0) 0.0 else 1.0)).toDouble
  }
}

// the values are written in their cells, rather than added as annotations
val data = Seq(
  Heatmap(visitors, days, hours).updated(
    colorscale = ColorScale.NamedScale("YlGnBu"),
    reversescale = true,
    texttemplate = "%{z:.0f}",
    textfont = Font(11),
    hovertemplate = "%{x} %{y}: %{z:.0f} visitors<extra></extra>",
    xgap = 2.0,
    ygap = 2.0
  )
)

val layout = Layout().updated(
  title = "Museum Visitors (Closed on Mondays)",
  yaxis = Axis().updated(autorange = true)
)

plot(data, layout)
```

## Shared Color Axis

```scala
val hours = (0 until 24).map(h => s"${h}h")
val days  = Seq("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

// simulated temperatures, in °C, of two cities, over a week
def temperatures(mean: Double, amplitude: Double, seed: Int) = {
  val random = new scala.util.Random(seed) // fixed seed, so that the plot is the same every time
  days.indices.map { d =>
    val dayMean = mean + 2 * random.nextGaussian()
    hours.indices.map(h => dayMean + amplitude * math.sin((h - 9) * math.Pi / 12))
  }
}

// both heatmaps use the color scale and bounds of the same color axis, so that their colors
// compare, and share its color bar
val data = Seq(
  Heatmap(temperatures(8.0, 5.0, 1), hours, days).updated(name = "Oslo", coloraxis = "coloraxis"),
  Heatmap(temperatures(24.0, 6.0, 2), hours, days).updated(
    name = "Seville",
    coloraxis = "coloraxis",
    xaxis = AxisReference.X2,
    yaxis = AxisReference.Y2
  )
)

val layout = Layout().updated(
  title = "Temperatures in Oslo and Seville",
  coloraxis = ColorAxis().updated(
    colorscale = ColorScale.NamedScale("RdBu"),
    colorbar = ColorBar().updated(ticksuffix = "°C", thickness = 15.0)
  ),
  grid = Grid().updated(rows = 1, columns = 2, pattern = Pattern.Independent),
  xaxis = Axis().updated(title = "Oslo", dtick = 6.0),
  xaxis2 = Axis().updated(title = "Seville", dtick = 6.0)
)

plot(data, layout)
```

## Cumulative Histogram

```scala
// simulated durations of 1,000 phone calls, in minutes
val durations = {
  // fixed seed, so that the plot is the same every time
  val random = new scala.util.Random(4)
  Seq.fill(1000)(-4 * math.log(1 - random.nextDouble()))
}

// for each duration, the share of the calls lasting longer: the bins are summed from the right
val data = Seq(
  Histogram(durations).updated(
    xbins = Bins(0.0, 20.0, 2.0),
    histnorm = HistNorm.Percent,
    cumulative = Cumulative(true).updated(direction = CumulativeDirection.Decreasing),
    texttemplate = "%{y:.0f}%",
    textposition = BarTextPosition.Outside,
    marker = Marker().updated(color = Color.RGB(25, 211, 243))
  )
)

val layout = Layout().updated(
  title = "Calls Lasting Longer Than…",
  xaxis = Axis().updated(title = "minutes", dtick = 2.0),
  yaxis = Axis().updated(ticksuffix = "%", range = (0.0, 110.0)),
  bargap = 0.05
)

plot(data, layout)
```

# Financial Charts
## Ohlc Chart

```scala
// simulated daily prices of a stock over six weeks of trading days, from Monday, March 3rd, 2025
val days = for {
  week <- 0 until 6
  day  <- 0 until 5
  dayOfMarch = 3 + 7 * week + day
} yield
  if (dayOfMarch <= 31) LocalDateTime(2025, 3, dayOfMarch, 0, 0, 0)
  else LocalDateTime(2025, 4, dayOfMarch - 31, 0, 0, 0)

val random = new scala.util.Random(5) // fixed seed, so that the plot is the same every time
val prices = days
  .scanLeft((0.0, 100.0)) { case ((_, previousClose), _) =>
    val open  = previousClose * (1 + 0.005 * random.nextGaussian())
    val close = open * (1 + 0.02 * random.nextGaussian())
    (open, close)
  }
  .tail
val highs = prices.map { case (open, close) => math.max(open, close) * (1 + 0.01 * random.nextDouble()) }
val lows  = prices.map { case (open, close) => math.min(open, close) * (1 - 0.01 * random.nextDouble()) }

val data = Seq(
  Ohlc(days, prices.map(_._1), highs, lows, prices.map(_._2)).updated(
    name = "ACME",
    increasing = DirectionStyle().updated(line = Line().updated(color = Color.RGB(0, 204, 150))),
    decreasing = DirectionStyle().updated(line = Line().updated(color = Color.RGB(239, 85, 59))),
    line = Line().updated(width = 2.0)
  )
)

val layout = Layout().updated(
  title = "ACME Stock Price",
  yaxis = Axis().updated(tickprefix = "$")
)

plot(data, layout)
```

## Split Violin Plot

```scala
val days = Seq("Thu", "Fri", "Sat", "Sun")

// simulated tips of lunch and dinner bills, by day
val random             = new scala.util.Random(11) // fixed seed, so that the plot is the same every time
def tips(mean: Double) = Seq.fill(80)(math.max(1.0, mean + 1.2 * random.nextGaussian()))

def sample(means: Seq[Double]) = {
  val pairs = days.zip(means).flatMap { case (day, mean) => tips(mean).map(day -> _) }
  (pairs.map(_._1), pairs.map(_._2))
}

// the lunch and dinner violins of each day share an axis, on their own side of it
val data = Seq(
  ("Lunch", ViolinSide.Negative, Color.RGB(99, 110, 250), Seq(2.6, 2.8, 3.0, 3.2)),
  ("Dinner", ViolinSide.Positive, Color.RGB(239, 85, 59), Seq(3.0, 3.1, 3.6, 3.8))
).map { case (meal, side, color, means) =>
  val (x, y) = sample(means)
  Violin(y, x).updated(
    name = meal,
    legendgroup = meal,
    scalegroup = meal,
    side = side,
    width = 0.9,
    line = Line().updated(color = color),
    meanline = MeanLine().updated(visible = true),
    points = BoxPoints.False
  )
}

val layout = Layout().updated(
  title = "Tips by Day and Meal",
  yaxis = Axis().updated(title = "tip ($)", zeroline = false),
  violinmode = BoxMode.Overlay
)

plot(data, layout)
```

# Contour Plots
## Basic Contour Plot

```scala
val xs = (-30 to 30).map(_ / 10.0)
val ys = (-20 to 20).map(_ / 10.0)

// two hills and a valley
def height(x: Double, y: Double) = {
  def bump(cx: Double, cy: Double, h: Double) =
    h * math.exp(-((x - cx) * (x - cx) + (y - cy) * (y - cy)))
  bump(-1.2, 0.4, 3) + bump(1.4, -0.3, 2) - bump(0.2, 1.4, 1.5)
}

val data = Seq(
  Contour(ys.map(y => xs.map(x => height(x, y))), xs, ys).updated(
    colorscale = ColorScale.NamedScale("Earth"),
    contours = Contours().updated(start = -1.5, end = 3.0, size = 0.25),
    colorbar = ColorBar().updated(title = Title.from(text = "Altitude", side = Side.Right))
  )
)

val layout = Layout().updated(title = "Two Hills and a Valley")

plot(data, layout)
```

## Contour Lines With Labels

```scala
val xs = (0 to 40).map(_ * math.Pi / 20)
val ys = (0 to 40).map(_ * math.Pi / 20)

// an egg-crate surface
val z = ys.map(y => xs.map(x => math.sin(x) * math.cos(y)))

val data = Seq(
  Contour(z, xs, ys).updated(
    ncontours = 12,
    // only lines, colored by level, and labelled with it
    contours = Contours().updated(
      coloring = ContourColoring.Lines,
      showlabels = true,
      labelfont = Font(12, "Inter, sans-serif", Color.StringColor("gray"))
    ),
    line = Line().updated(width = 2.0, smoothing = 0.85),
    colorscale = ColorScale.NamedScale("RdBu"),
    showscale = false
  )
)

val layout = Layout().updated(
  title = "sin(x) cos(y)",
  xaxis = Axis().updated(showgrid = false, zeroline = false),
  yaxis = Axis().updated(showgrid = false, zeroline = false)
)

plot(data, layout)
```

# Pie Charts
## Basic Pie Chart

```scala
// share of the electricity generated in a year, by source, in %
val sources = Seq(
  "Gas"     -> 38.0,
  "Nuclear" -> 18.0,
  "Coal"    -> 16.0,
  "Wind"    -> 14.0,
  "Solar"   -> 8.0,
  "Hydro"   -> 6.0
)

val data = Seq(
  Pie(sources.map(_._2), sources.map(_._1)).updated(
    textinfo = TextInfo(TextInfo.Label, TextInfo.Percent),
    hoverinfo = HoverInfo(HoverInfo.Label, HoverInfo.Value),
    // pull the renewable sources out of the pie
    pull = sources.map(s => if (Set("Wind", "Solar", "Hydro")(s._1)) 0.06 else 0.0),
    sort = false,
    direction = RotationDirection.Clockwise,
    marker = Marker().updated(line = Line().updated(color = Color.StringColor("white"), width = 2.0))
  )
)

val layout = Layout().updated(title = "Electricity Generation", showlegend = false)

plot(data, layout)
```

## Donut Chart

```scala
val platforms = Seq("Desktop", "Mobile", "Tablet")

// share of the visits of a website, by platform, in %, two years apart
val years = Seq(
  "2015" -> Seq(62.0, 30.0, 8.0),
  "2025" -> Seq(35.0, 59.0, 6.0)
)

// one donut per year, side by side, with the year in its hole
val data = years.zipWithIndex.map { case ((year, shares), idx) =>
  Pie(shares, platforms).updated(
    name = year,
    hole = 0.5,
    textinfo = TextInfo(TextInfo.Percent),
    hoverinfo = HoverInfo(HoverInfo.Label, HoverInfo.Percent, HoverInfo.Name),
    domain = Domain().updated(column = idx)
  )
}

val layout = Layout().updated(
  title = "Website Visits by Platform",
  grid = Grid().updated(rows = 1, columns = 2),
  annotations = years.zipWithIndex.map { case ((year, _), idx) =>
    Annotation().updated(
      text = year,
      x = 0.225 + idx * 0.55,
      y = 0.5,
      xref = Ref.Paper,
      yref = Ref.Paper,
      showarrow = false,
      font = Font(20)
    )
  }
)

plot(data, layout)
```
