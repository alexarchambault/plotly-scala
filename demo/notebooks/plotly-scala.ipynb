# Examples of plotly-scala with Almond

These are the examples of the [plotly-scala demo](https://alexarchambault.github.io/plotly-scala/).

```scala
//> using dep org.plotly-scala::plotly-almond:0.8.6-SNAPSHOT

// restrict the output height to avoid scrolling in output cells
repl.pprinter() = repl.pprinter().copy(defaultHeight = 3)

// to have the plots available without an internet connection, call init(offline=true) after this import
import plotly._, plotly.element._, plotly.layout._, plotly.Almond._
```

# Line Charts
## Basic Line Plot

```scala
val trace1 = Scatter(
  Seq(1, 2, 3, 4),
  Seq(10, 15, 13, 17)
)

val trace2 = Scatter(
  Seq(1, 2, 3, 4),
  Seq(16, 5, 11, 9)
)

val data = Seq(trace1, trace2)

plot(data)
```

## Line And Scatter Plot

```scala
val trace1 = Scatter(Seq(1, 2, 3, 4), Seq(10, 15, 13, 17))
  .withValues(mode = ScatterMode(ScatterMode.Markers))

val trace2 = Scatter(Seq(2, 3, 4, 5), Seq(16, 5, 11, 9))
  .withValues(mode = ScatterMode(ScatterMode.Lines))

val trace3 = Scatter(Seq(1, 2, 3, 4), Seq(12, 9, 15, 12))
  .withValues(mode = ScatterMode(ScatterMode.Lines, ScatterMode.Markers))

val data = Seq(trace1, trace2, trace3)

val layout = Layout().withValues(title = "Line and Scatter Plot")

plot(data, layout)
```

# Scatter Plots
## Categorical Dot Plot

```scala
val country = Seq(
  "Switzerland (2011)",
  "Chile (2013)",
  "Japan (2014)",
  "United States (2012)",
  "Slovenia (2014)",
  "Canada (2011)",
  "Poland (2010)",
  "Estonia (2015)",
  "Luxembourg (2013)",
  "Portugal (2011)"
)

val votingPop = Seq(
  40.0, 45.7, 52.0, 53.6, 54.1, 54.2, 54.5, 54.7, 55.1, 56.6
)

val regVoters = Seq(
  49.1, 42.0, 52.7, 84.3, 51.7, 61.1, 55.3, 64.2, 91.1, 58.9
)

val trace1 = Scatter(votingPop, country).withValues(
  mode = ScatterMode(ScatterMode.Markers),
  name = "Percent of estimated voting age population",
  marker = Marker().withValues(
    color = Color.RGBA(156, 165, 196, 0.95),
    line = Line().withValues(
      color = Color.RGBA(156, 165, 196, 1.0),
      width = 1.0
    ),
    symbol = Symbol.Circle(),
    size = 16
  )
)

val trace2 = Scatter(regVoters, country).withValues(
  mode = ScatterMode(ScatterMode.Markers),
  name = "Percent of estimated registered voters",
  marker = Marker().withValues(
    color = Color.RGBA(204, 204, 204, 0.95),
    line = Line().withValues(
      color = Color.RGBA(217, 217, 217, 1.0),
      width = 1.0
    ),
    symbol = Symbol.Circle(),
    size = 16
  )
)

val data = Seq(trace1, trace2)

val layout = Layout().withValues(
  title = "Votes cast for ten lowest voting age population in OECD countries",
  xaxis = Axis().withValues(
    showgrid = false,
    showline = true,
    linecolor = Color.RGB(102, 102, 102),
    titlefont = Font(Color.RGB(204, 204, 204)),
    tickfont = Font(Color.RGB(102, 102, 102)),
    autotick = false,
    dtick = 10.0,
    ticks = Ticks.Outside,
    tickcolor = Color.RGB(102, 102, 102)
  ),
  margin = Margin(
    l = 140,
    r = 40,
    b = 50,
    t = 80
  ),
  legend = Legend().withValues(
    font = Font(size = 10),
    yanchor = Anchor.Middle,
    xanchor = Anchor.Right
  ),
  width = 600,
  height = 400,
  paper_bgcolor = Color.RGB(254, 247, 234),
  plot_bgcolor = Color.RGB(254, 247, 234),
  hovermode = HoverMode.Closest
)

plot(data, layout)
```

## Grouped Scatter Plot

```scala
val countries = Seq("South Korea", "China", "Canada")

val data = Seq(
  "Gold"   -> Seq(24, 10, 9),
  "Silver" -> Seq(13, 15, 12),
  "Bronze" -> Seq(11, 8, 12)
).map { case (name, medals) =>
  Scatter(countries, medals).withValues(
    name = name,
    mode = ScatterMode(ScatterMode.Markers)
  )
}

val layout = Layout().withValues(
  title = "Grouped by Country",
  xaxis = Axis().withValues(title = "Country"),
  yaxis = Axis().withValues(title = "Medals"),
  scattermode = ScatterLayoutMode.Group,
  scattergap = 0.7
)

plot(data, layout)
```

# Bar Charts
## Basic Bar Chart

```scala
val data = Seq(
  Bar(
    Seq("giraffes", "orangutans", "monkeys"),
    Seq(20, 14, 23)
  )
)

plot(data)
```

## Grouped Bar Chart

```scala
val trace1 = Bar(Seq("giraffes", "orangutans", "monkeys"), Seq(20, 14, 23))
  .withValues(name = "SF Zoo")

val trace2 = Bar(Seq("giraffes", "orangutans", "monkeys"), Seq(12, 18, 29))
  .withValues(name = "LA Zoo")

val data = Seq(trace1, trace2)

val layout = Layout().withValues(barmode = BarMode.Group)

plot(data, layout)
```

## Bar Chart With Direct Labels

```scala
val xValue = Seq("Product A", "Product B", "Product C")

val yValue = Seq(20, 14, 23)

val trace1 = Bar(xValue, yValue).withValues(
  text = Seq("27% market share", "24% market share", "19% market share"),
  marker = Marker().withValues(
    color = Color.RGB(158, 202, 225),
    opacity = 0.6,
    line = Line().withValues(
      color = Color.RGB(8, 48, 107),
      width = 1.5
    )
  )
)

val data = Seq(trace1)

val annotations = xValue.zip(yValue).map { case (x, y) =>
  Annotation().withValues(
    x = x,
    y = y,
    text = y.toString,
    xanchor = Anchor.Center,
    yanchor = Anchor.Bottom,
    showarrow = false
  )
}

val layout = Layout().withValues(
  title = "January 2013 Sales Report",
  annotations = annotations
)

plot(data, layout)
```

## Customizing Individual Bar Colors

```scala
val defaultColor   = Color.RGBA(204, 204, 204, 1)
val highlightColor = Color.RGBA(222, 45, 38, 0.8)

val trace1 = Bar(
  Seq("Feature A", "Feature B", "Feature C", "Feature D", "Feature E"),
  Seq(20, 14, 23, 25, 22)
).withValues(
  marker = Marker().withValues(
    color = Seq(
      defaultColor,
      highlightColor,
      defaultColor,
      defaultColor,
      defaultColor
    )
  )
)

val data = Seq(trace1)

val layout = Layout().withValues(title = "Least Used Feature")

plot(data, layout)
```

## Waterfall Bar Chart

```scala
val xData = Seq(
  "Product Revenue",
  "Services Revenue",
  "Total Revenue",
  "Fixed Costs",
  "Variable Costs",
  "Total Costs",
  "Total"
)

val yData = Seq(400, 660, 660, 590, 400, 400, 340)

val textList = Seq("$430K", "$260K", "$690K", "$-120K", "$-200K", "$-320K", "$370K")

// Base

val trace1 = Bar(xData, Seq(0, 430, 0, 570, 370, 370, 0)).withValues(
  marker = Marker().withValues(color = Color.RGBA(1, 1, 1, 0.0))
)

// Revenue

val trace2 = Bar(xData, Seq(430, 260, 690, 0, 0, 0, 0)).withValues(
  marker = Marker().withValues(
    color = Color.RGBA(55, 128, 191, 0.7),
    line = Line().withValues(
      color = Color.RGBA(55, 128, 191, 1.0),
      width = 2.0
    )
  )
)

// Cost

val trace3 = Bar(xData, Seq(0, 0, 0, 120, 200, 320, 0)).withValues(
  marker = Marker().withValues(
    color = Color.RGBA(219, 64, 82, 0.7),
    line = Line().withValues(
      color = Color.RGBA(219, 64, 82, 1.0),
      width = 2.0
    )
  )
)

// Profit

val trace4 = Bar(xData, Seq(0, 0, 0, 0, 0, 0, 370)).withValues(
  marker = Marker().withValues(
    color = Color.RGBA(50, 171, 96, 0.7),
    line = Line().withValues(
      color = Color.RGBA(50, 171, 96, 1.0),
      width = 2.0
    )
  )
)

val data = Seq(trace1, trace2, trace3, trace4)

val annotations = xData.zip(yData).zip(textList).map { case ((x, y), text) =>
  Annotation().withValues(
    x = x,
    y = y,
    text = text,
    font = Font(
      family = "Arial",
      size = 14,
      color = Color.RGBA(245, 246, 249, 1)
    ),
    showarrow = false
  )
}

val layout = Layout().withValues(
  title = "Annual Profit 2015",
  barmode = BarMode.Stack,
  paper_bgcolor = Color.RGBA(245, 246, 249, 1),
  plot_bgcolor = Color.RGBA(245, 246, 249, 1),
  width = 600,
  height = 400,
  showlegend = false,
  annotations = annotations
)

plot(data, layout)
```

# Heatmaps
## Basic Heatmap

```scala
val data = Seq(
  Heatmap().withValues(
    z = Seq(
      Seq(1, 20, 30),
      Seq(20, 1, 60),
      Seq(30, 60, 1)
    ),
    colorscale = ColorScale.NamedScale("Portland")
  )
)

plot(data)
```

## Categorical Axis Heatmap

```scala
val data = Seq(
  Heatmap().withValues(
    z = Seq(
      Seq(1, null.asInstanceOf[Int], 30, 50, 1),
      Seq(20, 1, 60, 80, 30),
      Seq(30, 60, 1, -10, 20)
    ),
    x = Seq("Monday", "Tuesday", "Wednesday", "Thursday", "Friday"),
    y = Seq("Morning", "Afternoon", "Evening")
  )
)

plot(data)
```

## Custom Colorscale Heatmap

```scala
val data = Seq(
  Heatmap().withValues(
    z = Seq(
      Seq(10.0, 10.625, 12.5, 15.625, 20.0),
      Seq(5.625, 6.25, 8.125, 11.25, 15.625),
      Seq(2.5, 3.125, 5.0, 8.125, 12.5),
      Seq(0.625, 1.25, 3.125, 6.25, 10.625),
      Seq(0.0, 0.625, 2.5, 5.625, 10.0)
    ),
    colorscale = ColorScale.CustomScale(
      Seq(
        (0, Color.RGB(166, 206, 227)),
        (0.25, Color.RGB(31, 120, 180)),
        (0.45, Color.RGB(178, 223, 138)),
        (0.65, Color.RGB(51, 160, 44)),
        (0.85, Color.RGB(251, 154, 153)),
        (1, Color.RGB(227, 26, 28))
      )
    )
  )
)

plot(data)
```

## Annotated Heatmap

```scala
val x = Seq("A", "B", "C", "D", "E");
val y = Seq("W", "X", "Y", "Z");
val z = Seq(
  Seq(0.00, 0.00, 0.75, 0.75, 0.00),
  Seq(0.00, 0.00, 0.75, 0.75, 0.00),
  Seq(0.75, 0.75, 0.75, 0.75, 0.75),
  Seq(0.00, 0.00, 0.00, 0.75, 0.00)
)

val data = Seq(
  Heatmap(z, x, y).withValues(
    showscale = false,
    colorscale = ColorScale.CustomScale(
      Seq(
        (0, Color.StringColor("#3D9970")),
        (1, Color.StringColor("#001f3f"))
      )
    )
  )
)

val layout = Layout().withValues(
  title = "Annotated Heatmap",
  xaxis = Axis().withValues(ticks = Ticks.Empty, side = Side.Top),
  yaxis = Axis().withValues(ticks = Ticks.Empty, ticksuffix = " "),
  annotations = for {
    (xv, xi) <- x.zipWithIndex
    (yv, yi) <- y.zipWithIndex
  } yield Annotation().withValues(
    x = xv,
    y = yv,
    xref = Ref.Axis(AxisReference.X1),
    yref = Ref.Axis(AxisReference.Y1),
    showarrow = false,
    text = z(yi)(xi).toString,
    font = Font(Color.StringColor("white"))
  )
)

plot(data, layout)
```

# Histogram
## Basic Histogram

```scala
val categoryCount = 50

val indices    = 0 until categoryCount
val categories = indices.map(i => s"name-$i")
val values     = {
  // fixed seed, so that the plot is the same every time
  val random = new scala.util.Random(42)
  indices.map(_ => random.nextDouble())
}

val data = Seq(Histogram(values, categories))

val layout = Layout().withValues(
  title = "Basic Histogram",
  showlegend = false,
  height = 400,
  width = 600
)

plot(data, layout)
```

## Styled Basic Histogram

```scala
val categoryCount = 50

val indices    = 0 until categoryCount
val categories = indices.map(i => s"name-$i")
val values     = {
  // fixed seed, so that the plot is the same every time
  val random = new scala.util.Random(42)
  indices.map(_ => random.nextDouble())
}

val data = Seq(
  Histogram(values, categories).withValues(
    marker = Marker().withValues(color = Color.StringColor("#004A72")),
    hovertext = categories.map(c => s"$c with hover text")
  )
)

val xAxis = Axis().withValues(
  range = (0d, 2d),
  rangeslider = RangeSlider()
)

val yAxis = Axis().withValues(
  title = "Count",
  fixedrange = true,
  tickformat = ".1f"
)

val layout = Layout().withValues(
  xaxis = xAxis,
  yaxis = yAxis,
  title = "Styled Basic Histogram",
  showlegend = false,
  height = 400,
  width = 600
)

plot(data, layout)
```
